```python
import numpy as np
import csv

np.random.seed(42)
n = 120

# IDs patients
ID_patient = ["#{:04d}".format(i+1) for i in range(n)]

# Variables cliniques
age = np.random.randint(45, 85, n)

sexe = np.random.choice(["M", "F"], n)

antecedents = np.random.choice([0, 1], n, p=[0.7, 0.3])

depression = np.random.choice([0, 1], n, p=[0.6, 0.4])

ethnicite = np.random.choice(
    ["Europe", "Afrique", "Asie", "Amerique_Nord", "Amerique_Sud"],
    n
)

# Neuroanatomie

substance_noire = np.round(
    np.clip(np.random.normal(1.2, 0.2, n), 0.5, None),
    2
)

noyau_caudé = np.round(
    np.clip(np.random.normal(2.0, 0.3, n), 0.5, None),
    2
)

# Biomarqueurs

BHB = np.round(
    np.clip(np.random.normal(0.8, 0.4, n), 0.1, None),
    2
)

glycine = np.round(
    np.clip(np.random.normal(2.5, 0.6, n), 0.1, None),
    2
)

pyruvate = np.round(
    np.clip(np.random.normal(1.0, 0.3, n), 0.1, None),
    2
)

serine = np.round(
    np.clip(np.random.normal(1.8, 0.5, n), 0.1, None),
    2
)

betaine = np.round(
    np.clip(np.random.normal(1.2, 0.4, n), 0.1, None),
    2
)

# Score UPDRS

score = (
    0.03 * age +
    0.8 * depression +
    0.6 * antecedents -
    1.2 * substance_noire -
    0.8 * noyau_caudé +
    0.4 * BHB +
    0.2 * glycine +
    0.3 * pyruvate +
    0.2 * serine -
    0.1 * betaine +
    np.random.normal(0, 1, n)
)

# Normalisation score 0–4
score = (
    (score - np.min(score)) /
    (np.max(score) - np.min(score))
) * 4

score = np.round(score, 2)

# Classification UPDRS

UPDRS = []

for s in score:
    if s < 0.8:
        UPDRS.append(0)
    elif s < 1.6:
        UPDRS.append(1)
    elif s < 2.4:
        UPDRS.append(2)
    elif s < 3.2:
        UPDRS.append(3)
    else:
        UPDRS.append(4)

# Export CSV

with open("parkinson_120.csv", "w", newline="") as f:

    writer = csv.writer(f)

    # Entête
    writer.writerow([
        "ID_patient",
        "age", "sexe", "antecedents",
        "ethnicite", "depression",
        "substance_noire", "noyau_caudé",
        "BHB", "glycine", "pyruvate",
        "serine", "betaine",
        "UPDRS"
    ])

    # Données
    for i in range(n):

        writer.writerow([
            ID_patient[i],
            age[i],
            sexe[i],
            antecedents[i],
            ethnicite[i],
            depression[i],
            substance_noire[i],
            noyau_caudé[i],
            BHB[i],
            glycine[i],
            pyruvate[i],
            serine[i],
            betaine[i],
            UPDRS[i]
        ])

print("Dataset généré : parkinson_120.csv")
```
